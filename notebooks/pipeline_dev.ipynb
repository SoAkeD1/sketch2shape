# Pipeline dev
Runs each stage on your own sketch. Set `SKETCH_PATH` first. Run from `notebooks/` with `src` importable.

```python
import sys
sys.path.insert(0, '../src')
from sketch2shape import background, image_to_3d, preprocess, sketch_to_image
from sketch2shape.config import load_config
from sketch2shape.device import pick_device

SKETCH_PATH = ''  # path to a sketch you drew
PROMPT = None     # optional
assert SKETCH_PATH, 'Set SKETCH_PATH to a sketch you drew'
cfg = load_config()
device, dtype = pick_device(cfg)
device, dtype
```

```python
prepared = preprocess.prepare(SKETCH_PATH, cfg)
display(prepared.clean, prepared.control, prepared.mask)
```

```python
image = sketch_to_image.generate(prepared.control, PROMPT, cfg, device, dtype)
image
```

```python
rgba, foreground_mask = background.isolate(image, cfg)
rgba
```

```python
recon = image_to_3d.reconstruct(rgba, cfg, device)
recon.mesh.extents, len(recon.mesh.faces), recon.textured, recon.clean_stats
```
